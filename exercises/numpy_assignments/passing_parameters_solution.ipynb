# Exercise: Side effects of calling functions

Below you'll find three functions that add a given value twice to the end of a list.

Read the code and figure out what the initial list will look like after calling the function, then call the function and compare with your intuition.

```python
def augment_twice_A(a_list, val):
    """Put `val` at the end of `a_list` twice."""
    a_list.append(val)
    a_list.append(val)
    return a_list

def augment_twice_B(a_list, val):
    """Put `val` at the end of `a_list` twice."""
    a_list = a_list + [val, val]
    return a_list
    
def augment_twice_C(a_list, val):
    """Put `val` at the end of `a_list` twice."""
    a_list = list(a_list)
    a_list.append(val)
    a_list.append(val)
    return a_list
    
```

```python
numbers = [1, 2, 3]
result = augment_twice_A(numbers, 4)
print(numbers, result)
```

```python
numbers = [1, 2, 3]
result = augment_twice_B(numbers, 4)
print(numbers, result)
```

```python
numbers = [1, 2, 3]
result = augment_twice_C(numbers, 4)
print(numbers, result)
```
