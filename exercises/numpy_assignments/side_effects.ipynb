## A. Taking the temperature

```python
def record_temperature(patient_id, temp, readings={}):
    """Record a temperature reading for a patient."""
    readings.setdefault(patient_id, []).append(temp)
    return readings
```

```python
# Case 1: we have an initial log
patient_temperatures = {'Alice': [37.2], 'Bob': [39.5]}
patient_temperatures = record_temperature('Bob', 20.1, readings=patient_temperatures)
patient_temperatures = record_temperature('Alice', 38.9, readings=patient_temperatures)
```

```python
# Guess the result of this print statement before executing it!
print(patient_temperatures)
```

```python
# Case 1: we rely on the default argument
alice_temperatures = record_temperature('Alice', 37.2)
alice_temperatures = record_temperature('Alice', 41.4, readings=alice_temperatures)
alice_temperatures = record_temperature('Alice', 38.6, readings=alice_temperatures)

bob_temperatures = record_temperature('Bob', 38.9)
bob_temperatures = record_temperature('Bob', 36.1, readings=bob_temperatures)
bob_temperatures = record_temperature('Bob', 37.4, readings=bob_temperatures)
```

```python
# Guess the result of this print statement before executing it!
print(alice_temperatures)
```

## B. Let's sort it out (OPTIONAL)

```python
def oldest_n_patients(ages, n=3):
    """Return the ages of the n oldest patients."""
    ages.sort()
    return ages[-n:]
```

```python
patient_ages = [58, 77, 72, 71, 79]
oldest = oldest_n_patients(patient_ages, n=3)
```

```python
# Guess the result of this print statement before executing it!
print(oldest)
```
